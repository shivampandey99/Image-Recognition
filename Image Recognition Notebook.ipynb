# Image Recognition using MobileNetV2

## Project Overview
This notebook demonstrates image classification using the pretrained MobileNetV2 model. It covers loading the model, preprocessing an input image, running inference, and displaying the top predictions.

## Problem Statement
Build an image recognition pipeline that classifies an uploaded image using a pretrained deep learning model trained on ImageNet.

## Objectives
- Load a pretrained MobileNetV2 model.
- Preprocess an input image.
- Perform image classification.
- Display the Top-5 predictions.

```
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions
from tensorflow.keras.utils import img_to_array
```

## Load the Pretrained Model

```
# Load MobileNetV2 with pretrained ImageNet weights
model = MobileNetV2(weights="imagenet")
print("Model loaded successfully.")
```

## Load an Image
Replace the image path with your own test image.

```
image_path = "sample.jpg"  # Change this path

image = Image.open(image_path).convert("RGB")

plt.figure(figsize=(6,6))
plt.imshow(image)
plt.axis("off")
plt.title("Input Image")
plt.show()
```

## Preprocess the Image

```
# Resize to MobileNetV2 input size
resized = image.resize((224,224))

# Convert image to array
image_array = img_to_array(resized)

# Add batch dimension
image_batch = np.expand_dims(image_array, axis=0)

# Apply MobileNetV2 preprocessing
processed = preprocess_input(image_batch)
```

## Perform Image Recognition

```
predictions = model.predict(processed, verbose=0)
decoded = decode_predictions(predictions, top=5)[0]

print("Top 5 Predictions")
print("-"*45)
print(f"{'Rank':<6}{'Class':<30}{'Confidence'}")

for i, (_, label, prob) in enumerate(decoded, start=1):
    print(f"{i:<6}{label.replace('_',' '):<30}{prob*100:.2f}%")
```

## Conclusion

This notebook demonstrated an end-to-end image recognition workflow using the pretrained MobileNetV2 model. The notebook focuses solely on experimentation and inference. Deployment code belongs in `app.py`.
